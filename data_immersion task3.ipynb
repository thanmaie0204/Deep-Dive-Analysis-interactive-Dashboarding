```python
import pandas as pd

df = pd.read_csv("final_eda_dataset.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
# Core KPIs

total_revenue = df["Total_Sales"].sum()
total_quantity = df["Quantity"].sum()
total_customers = df["Customer_ID"].nunique()
average_order_value = df["Total_Sales"].mean()

print("Total Revenue:", total_revenue)
print("Total Quantity Sold:", total_quantity)
print("Total Customers:", total_customers)
print("Average Order Value:", average_order_value)
```

```python
# Deep-Dive Analysis: Category Performance

category_sales = df.groupby("Category")["Total_Sales"].sum().sort_values(ascending=False)

print(category_sales)
```

```python
category_sales.plot(kind="bar", figsize=(8,5), title="Total Sales by Category")
```

```python
# City-wise Sales Analysis

city_sales = df.groupby("City")["Total_Sales"].sum().sort_values(ascending=False)

print(city_sales)
```

```python
city_sales.plot(
    kind="bar",
    figsize=(8,5),
    title="Total Sales by City"
)
```

```python
# Customer Segmentation

customer_sales = df.groupby("Customer_ID")["Total_Sales"].sum()

def segment_customer(sales):
    if sales >= customer_sales.quantile(0.75):
        return "High Value"
    elif sales >= customer_sales.quantile(0.50):
        return "Medium Value"
    else:
        return "Low Value"

customer_segment = customer_sales.apply(segment_customer)

print(customer_segment.value_counts())
```

```python
customer_segment.value_counts().plot(
    kind="bar",
    figsize=(8,5),
    title="Customer Segmentation"
)
```
